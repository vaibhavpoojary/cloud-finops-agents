```python
import os, glob, getpass

home = os.path.expanduser("~")
candidates = []
# common GPT4All model locations (Windows + Linux)
paths = [
    os.path.join(home, ".gpt4all", "models"),
    os.path.join(home, ".cache", "gpt4all"),
    os.path.join(home, "AppData", "Local", "gpt4all", "models"),
    os.path.join(os.getcwd(), "models"),
]

# also search whole home if nothing found in common places (safe small search)
for p in paths:
    if os.path.isdir(p):
        candidates += glob.glob(os.path.join(p, "*.*"))

# fallback: search home (can be slow — you can skip if you don't want it)
if not candidates:
    candidates += glob.glob(os.path.join(home, "**", "*.gguf"), recursive=True)
    candidates += glob.glob(os.path.join(home, "**", "*.bin"), recursive=True)

candidates = sorted(set(candidates))
print(f"Found {len(candidates)} model files:")
for f in candidates:
    print(" -", f)
```
